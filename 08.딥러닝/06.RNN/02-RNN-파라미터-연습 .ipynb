```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Input
```

```python
x = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6],
])


# x = np.array([
#     [
#         [1],
#         [2],
#         [3]
#     ],
#     [
#         [2],
#         [3],
#         [4]
#     ],
#     [
#         [3],
#         [4],
#         [5]
#     ],
#     [
#         [4],
#         [5],
#         [6]
#     ]
# ])

y = np.array([4,5,6,7])

x = x.reshape(-1,3,1) # 굳이 위에와 같이 데이터를 바꾸지 않아도 reshape를 사용하여 변환시킬 수 있음 

print(x)
print(x.shape)  

model = Sequential([
    Input(shape=(3,1)),
    SimpleRNN(2),
    Dense(1)
])

model.summary()

model.compile(
    optimizer="adam",
    loss="mse"    
)

model.fit(
    x,
    y,
    epochs=300,
    verbose=1
)
```

```python
# 예측
x_pred = np.array([[5,6,7]]).reshape(-1,3,1)
# pred = model.predict(np.array([[[5],[6],[7]]]))
pred = model.predict(x_pred)

# 두 개의 출력의 값이 똑같음

print(pred)
```

```python
# 주식 종가 예측

x = np.array([
    [50000, 51000, 52000, 53000],
    [51000, 52000, 53000, 54000],
    [52000, 53000, 54000, 55000]
])

# 다음 날 종가
y = np.array([
    54000,
    55000,
    56000
])

x = x.reshape(-1,4,1)  # x = x.reshape(3,4,1)으로 해도 되긴 함, -1은 알아서 자동으로 맞춰줌 => 잘 모르겠으면 -1 하는 것이 좋음
print(x)
print(x.shape)

# 모델 생성

# data_size = 3
# time_steps = 4
# features = 1 ( 종가만 사용)

model = Sequential([
    Input(shape=(4,1)),
    SimpleRNN(4, activation="relu"),  # 시간과 관련하여 예측하는 건 RNN을 대부분 기본적으로 사용함
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.summary()

model.fit(
    x,
    y,
    epochs=500,
    verbose=1
)

#[53000,54000,55000,56000]
```

```python
x_preb = np.array([[53000,54000,55000,56000]]).reshape(-1,4,1)
preb = model.predict(x_preb)
print("다음 날 예상 종가 :",preb)
```

```python
# Feature 1 : 온도 (Temperature)
# Feature 2 : 습도 (Humidity)

# 즉, 3일간의 온도와 습도를 보고 다음 날의 온도를 예측하시오

# [온도, 습도]
x = np.array([
    [[20,60],[21,61],[22,62]],
    [[23,63],[24,64],[25,65]],
    [[26,66],[27,67],[28,68]]
])

# 다음 날 온도
y = np.array([
    23,
    26,
    29
])

model = Sequential([
    Input(shape=(3,2)),
    SimpleRNN(8,activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.summary()

model.fit(
    x,
    y,
    epochs=500,
    verbose=1
)
# 예측하시오
#    [[29,69],
#      [30,70],
#      [31,71]]
```

```python
pred = model.predict(np.array([[[29,69],[30,70],[31,71]]]))
print("예상 온도 : ", pred)
```

```python
# ==========================================
# 1. 데이터 준비
# ==========================================

# [키(cm), 몸무게(kg), 나이]
x = np.array([
    [[170,60,20], [171,61,21]],
    [[180,70,30], [181,71,31]],
    [[190,80,40], [191,81,41]],
    [[200,90,50], [201,91,51]]
])

# 다음 해 키
y = np.array([
    [172,62],
    [182,72],
    [192,82],
    [202,92]
])

x = x.reshape(-1,2,3)
print("===== 입력 데이터 =====")
print(x)

print("\nShape :", x.shape)


model = Sequential([
    Input(shape=(2,3)),
    SimpleRNN(16, activation="relu"),
    Dense(2)  # 예측하는 컬럼이 2개 이므로 2가 됨
])

model.summary()

model.compile(
    optimizer="adam",
    loss="mse"
)

model.fit(
    x,
    y,
    epochs=500,
    verbose=1
)

```

```python
x_pred = np.array([
    [[175,65,25],
    [176,66,26]]    
])


pred = model.predict(x_pred)
print(pred)
```
