```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
import numpy as np

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    0,
    1,
    1,
    0
])
```

```python
model = Sequential([
    Input(shape=(2,)),
    Dense(4,activation="relu"),  
    Dense(1,activation="sigmoid")
])

# 히든층의 activation은 relu로 거의 정해져있음 , 은닉층을 추가하지 않으면 어떠한 선으로도 분류하지 못함 => loss가 줄어들지 않음
# 은닉층이 필요한 이유 : 추가됨으로써 좌표의 위치가 바뀜, 공간을 변형시킴, 특성을 변경시킴 => 다 같은 의미
# XOR의 문제는 은닉층으로 해결

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.fit(X,y,epochs=3000)
```

```python
pred = model.predict(X)

print(pred)
```
