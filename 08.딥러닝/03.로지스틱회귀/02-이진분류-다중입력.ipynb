```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
import numpy as np

# 키 + 몸무게 => 비만

X = np.array([
    [170,60],
    [175,65],
    [180,70],
    [170,90],
    [175,95],
    [180,100]
])

y = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])
```

```python
# 모델 생성

model = Sequential([
    Input(shape=(2,)),
    Dense(1,activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 학습
model.fit(X,y,epochs=1000,verbose=1)
```

```python
pred = model.predict(np.array([[172,88]]))  # 넘파이 배열로 사용하는 이유는 수학적 행렬 계산이 가능하기 때문에

print(pred)

if pred[0][0] >= 0.5:
    print("비만")
else:
    print("정상")
```

```python
model.summary()
```

```python
# [키, 몸무게, 허리둘레]

X = np.array([
    [170, 60, 75],
    [175, 65, 78],
    [180, 70, 80],
    [170, 90, 100],
    [175, 95, 105],
    [180, 100, 110]
])

# 정상=0, 비만=1

y = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])

# [172, 88, 98]

model = Sequential([
    Input(shape=(3,)),
    Dense(1,activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 학습
model.fit(X,y,epochs=1000,verbose=1)
```

```python
pred = model.predict(np.array([[172,88,98]]))

print(pred)

if pred[0][0] >= 0.5:
    print("비만")
else:
    print("정상")
```
