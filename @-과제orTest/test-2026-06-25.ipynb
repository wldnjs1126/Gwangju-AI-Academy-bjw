# 1. SQL 정리

## 1-1. RDBMS

**RDBMS**는 Relational Database Management System의 약자이며, 한국어로는 **관계형 데이터베이스 관리 시스템**이다.

데이터를 **행과 열로 이루어진 테이블 형태**로 저장하고, 여러 테이블을 관계로 연결해서 관리하는 시스템이다.

### 시험 답안

```text
RDBMS는 관계형 데이터베이스 관리 시스템으로, 데이터를 행과 열로 이루어진 테이블 형태로 저장하고 여러 테이블 간 관계를 이용해 데이터를 관리하는 시스템이다.
```

---

## 1-2. DML / DDL / DCL

| 구분 | 뜻 | 의미 | 대표 명령어 |
|---|---|---|---|
| DML | Data Manipulation Language | 데이터 조작어 | SELECT, INSERT, UPDATE, DELETE |
| DDL | Data Definition Language | 데이터 정의어 | CREATE, ALTER, DROP |
| DCL | Data Control Language | 데이터 제어어 | GRANT, REVOKE |

### 시험 답안

```text
DML은 데이터 조작어로, 테이블 안의 데이터를 조회, 추가, 수정, 삭제할 때 사용하는 SQL 명령어이다.
DDL은 데이터 정의어로, 테이블이나 데이터베이스의 구조를 생성, 변경, 삭제할 때 사용하는 SQL 명령어이다.
DCL은 데이터 제어어로, 데이터베이스 사용 권한을 부여하거나 회수할 때 사용하는 SQL 명령어이다.
```

## 1-6. WHERE 조건절

`where`는 조건에 맞는 행만 조회할 때 사용한다.

```sql
select *
from emp
where deptno = 10;
```

급여가 3000 이상:

```sql
select *
from emp
where sal >= 3000;
```

직업이 CLERK가 아닌 사원:

```sql
select *
from emp
where job <> 'CLERK';
```

| 연산자 | 의미 |
|---|---|
| `=` | 같다 |
| `<>` | 같지 않다 |
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |

## 1-7. IN

## 1-10. LIKE, %, _

`like`는 문자열 패턴 검색이다.

```sql
select *
from emp
where ename like 'J%';
```

뜻: 이름이 J로 시작하는 사원 조회

```sql
select *
from emp
where ename like '%J%';
```

뜻: 이름에 J가 포함된 사원 조회

```sql
select *
from emp
where ename like '_A%';
```

뜻: 두 번째 글자가 A인 사원 조회

| 기호 | 의미 |
|---|---|
| `%` | 0개 이상의 아무 문자 |
| `_` | 정확히 1개의 아무 문자 |
| `J%` | J로 시작 |
| `%J%` | J 포함 |
| `_A%` | 두 번째 글자가 A |

## 1-13. GROUP BY

`group by`는 같은 값끼리 묶어서 집계할 때 사용한다.

부서별 급여 평균:

```sql
select deptno, avg(sal)
from emp
group by deptno;
```

## 1-14. HAVING

`having`은 그룹화된 결과에 조건을 줄 때 사용한다.

```sql
select deptno, avg(sal) as 평균
from emp
group by deptno
having avg(sal) >= 2000;
```

뜻:

```text
부서별 평균 급여를 구한 뒤, 평균 급여가 2000 이상인 부서만 조회한다.
```

| 구분 | 사용 위치 | 의미 |
|---|---|---|
| `where` | group by 전 | 원본 행에 조건 |
| `having` | group by 후 | 그룹 결과에 조건 |

## 1-15. JOIN

JOIN은 두 테이블을 연결해서 조회하는 것이다.

기본 구조:

```sql
select 컬럼명
from 테이블1
join 테이블2
on 연결조건;
```

예시:

```sql
select ename, dept.deptno, dname
from emp
join dept
on emp.deptno = dept.deptno;
```

뜻:

```text
emp 테이블과 dept 테이블을 부서번호 기준으로 연결하여 사원이름, 부서번호, 부서명을 출력한다.
```

---

## 1-16. LEFT OUTER JOIN

왼쪽 테이블은 무조건 다 출력하고, 오른쪽 테이블에 매칭되는 값이 있으면 붙인다. 없으면 NULL로 채운다.

```sql
select *
from dept
left outer join emp
on emp.deptno = dept.deptno;
```

### 시험 답안

```text
LEFT OUTER JOIN은 왼쪽 테이블의 모든 데이터를 출력하고, 오른쪽 테이블의 일치하는 데이터가 없으면 NULL로 채운다.
```

---

## 1-17. SELF JOIN

SELF JOIN은 같은 테이블을 자기 자신과 조인하는 것이다.

```sql
select *
from emp e
join emp m
on e.mgr = m.empno;
```

여기서 `e`는 사원, `m`은 매니저 역할이다.

### 시험 답안

```text
SELF JOIN은 같은 테이블을 서로 다른 별칭으로 두 번 사용하여 자기 자신과 조인하는 방식이다.
```

## 1-18. 서브쿼리

서브쿼리는 쿼리 안에 들어가는 쿼리이다.

```sql
select 컬럼명
from 테이블명
where 컬럼명 = (
    select 컬럼명
    from 테이블명
    where 조건
);
```

예시:

```sql
select empno, ename
from emp
where job = (
    select job
    from emp
    where empno = 7900
);
```

뜻:

```text
사원번호 7900번 사원의 직업과 같은 직업을 가진 사원의 사원번호와 이름을 조회한다.
```

평균 급여보다 많이 받는 사원:

```sql
select empno, ename, sal
from emp
where sal > (
    select avg(sal)
    from emp
);
```

데이터 추가:

NumPy는 파이썬에서 배열을 다루기 위한 대표 패키지이다. 파이썬 기본 리스트보다 빠르고, 대량의 수치 데이터를 효율적으로 처리할 수 있다.

## 2-5. 인덱싱과 슬라이싱

인덱싱은 한 개의 값을 위치로 가져오는 것이다.

```python
arr = np.array([0, 1, 2, 3, 4])
print(arr[0])
print(arr[2])
print(arr[-1])
```

2차원 배열:

```python
arr = np.array([
    [0, 1, 2],
    [3, 4, 5]
])

print(arr[0, 0])
print(arr[1, 2])
```

슬라이싱 기본 구조:

```python
배열[start:end:step]
```

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70])

print(arr[4:])
print(arr[1:4])
print(arr[::2])
```

2차원 슬라이싱:

```python
arr = np.array([
    [0, 1, 2, 3],
    [4, 5, 6, 7]
])

print(arr[0, :])
print(arr[1, :])
print(arr[:, 1])
```

| 코드 | 의미 |
|---|---|
| `arr[0, :]` | 첫 번째 행 전체 |
| `arr[1, :]` | 두 번째 행 전체 |
| `arr[:, 1]` | 두 번째 열 전체 |

# 3. Pandas 정리

## 3-1. Series

Series는 Pandas의 1차원 자료구조이다.

```text
행번호 + 인덱스 + 데이터
```

### 시험 답안

## 3-15. apply

`apply()`는 함수 하나를 열 전체에 적용하는 것이다.

```python
def add_s(age):
    return str(age) + "세"

df['나이'] = df['나이'].apply(add_s)
```

```python
def remove_s(age_str):
    return int(str(age_str).replace('세', ''))

df['나이'] = df['나이'].apply(remove_s)
```

# 7. 통계 분석 정리

## 7-1. 기술통계와 추측통계

| 구분 | 의미 |
|---|---|
| 기술통계 | 수집된 데이터를 정리, 요약, 설명 |
| 추측통계 | 표본을 이용해 모집단을 추정하거나 검정 |

### 시험 답안

```text
기술통계는 수집된 데이터를 표, 그래프, 수치 요약으로 정리하고 설명하는 통계이고, 추측통계는 일부 표본 데이터를 이용하여 모집단의 특성을 추정하거나 가설을 검정하는 통계이다.
```

---

## 7-2. 평균, 편차, 분산, 표준편차

| 개념 | 의미 |
|---|---|
| 평균 | 전체 데이터를 대표하는 중심값 |
| 편차 | 각 값 - 평균 |
| 분산 | 편차를 제곱한 값들의 평균 |
| 표준편차 | 분산에 루트를 씌운 값 |

분산에 루트를 씌우는 이유:

```text
분산은 편차를 제곱해서 원래 단위보다 커지므로, 원래 데이터 단위에 가깝게 만들기 위해 루트를 씌운다.
```

---

## 7-3. 정규분포와 표준정규분포

정규분포는 종 모양의 분포이다.

```text
N(평균, 분산)
```

표준정규분포는 평균이 0, 표준편차가 1인 정규분포이다.

```text
N(0, 1)
```

### 시험 답안

```text
표준정규분포는 평균이 0이고 표준편차가 1인 정규분포이며, 서로 다른 정규분포를 비교하기 위해 표준화한 분포이다.
```

---

## 7-4. 귀무가설과 대립가설

| 구분 | 의미 |
|---|---|
| 귀무가설 H0 | 차이가 없다, 효과가 없다, 관계가 없다 |
| 대립가설 H1 | 차이가 있다, 효과가 있다, 관계가 있다 |

---

## 7-5. p-value

p-value는 귀무가설이 맞다고 가정했을 때, 현재와 같은 결과가 나올 확률이다.

판단 기준:

```text
p-value > 0.05 → 귀무가설 유지
p-value < 0.05 → 귀무가설 기각
```

### 시험 답안

```text
p-value가 0.05보다 작으면 통계적으로 유의하다고 보고 귀무가설을 기각한다.
```

---

## 7-6. Shapiro 정규성 검정

Shapiro 검정은 데이터가 정규분포를 따르는지 확인하는 검정이다.

```python
from scipy import stats

st, pv = stats.shapiro(data)
```

가설:

```text
귀무가설: 데이터는 정규분포를 따른다.
대립가설: 데이터는 정규분포를 따르지 않는다.
```

판단:

```text
p-value > 0.05 → 정규분포를 따른다고 봄
p-value < 0.05 → 정규분포를 따르지 않는다고 봄
```

---

## 7-7. 일표본 t-test

한 집단의 평균이 특정 기준값과 같은지 확인한다.

```python
from scipy import stats

data = [200, 210, 190, 205, 198]
mu = 200

t_test = stats.ttest_1samp(data, mu)
print(t_test)
```

구조:

```python
stats.ttest_1samp(표본데이터, 기준평균)
```

### 시험 답안

```text
일표본 t-검정은 한 집단의 평균이 특정 기준값과 통계적으로 차이가 있는지 검정하는 방법이다.
```

---

## 7-8. 독립표본 t-test

서로 다른 두 집단의 평균을 비교한다.

```python
st, pv = stats.ttest_ind(boys, girls)
```

구조:

```python
stats.ttest_ind(집단1, 집단2)
```

### 시험 답안

```text
독립표본 t-검정은 서로 독립적인 두 집단의 평균 차이가 통계적으로 유의한지 확인하는 검정이다.
```

---

## 7-9. 대응표본 t-test

같은 집단을 전후로 비교한다.

```python
st, pv = stats.ttest_rel(before, after)
```

구조:

```python
stats.ttest_rel(전, 후)
```

### 시험 답안

```text
대응표본 t-검정은 같은 대상에게 두 번 측정한 값의 평균 차이를 비교하는 검정이다.
```

---

## 7-10. t-test 비교표

| 검정 | 비교 대상 | 데이터 관계 | 함수 |
|---|---|---|---|
| 일표본 t-test | 한 집단 vs 기준값 | 한 집단 | `ttest_1samp()` |
| 독립표본 t-test | 두 집단 평균 | 서로 다른 집단 | `ttest_ind()` |
| 대응표본 t-test | 전후 평균 | 같은 집단 | `ttest_rel()` |

---

## 7-11. 카이제곱 검정

카이제곱 검정은 범주형 변수 2개의 관계를 확인한다.

```text
범주형 × 범주형
```

순서:

```text
crosstab() → chi2_contingency()
```

코드:

```python
import pandas as pd
from scipy.stats import chi2_contingency

table = pd.crosstab(df["성별"], df["합격"])
chi2, p, _, _ = chi2_contingency(table)

print(chi2)
print(p)
```

### 시험 답안

```text
카이제곱 검정은 두 범주형 변수 사이에 관련성이 있는지 확인하는 검정이다.
```

`_`의 의미:

```text
필요하지 않은 반환값을 받기 위해 사용하는 변수명이다.
```

---

## 7-12. ANOVA

ANOVA는 세 집단 이상의 평균 차이를 비교한다.

```python
f_stat, p_value = stats.f_oneway(A, B, C)
```

### 시험 답안

```text
ANOVA는 세 개 이상의 집단 평균 차이가 통계적으로 유의한지 확인하는 분산분석이다.
```

---

## 7-13. 검정 선택표

| 상황 | 검정 |
|---|---|
| 데이터가 정규분포인지 확인 | Shapiro |
| 한 집단 평균 vs 기준값 | 일표본 t-test |
| 서로 다른 두 집단 평균 비교 | 독립표본 t-test |
| 같은 집단 전후 평균 비교 | 대응표본 t-test |
| 범주형 × 범주형 관계 확인 | 카이제곱 검정 |
| 세 집단 이상 평균 비교 | ANOVA |
