```python
# 인덱스 컨트롤하기
import pandas as pd
import numpy as np

# 딕셔너리로 생성 시 자동으로 키가 시리즈의 인덱스로 들어감
# 딕셔너리의 구조는 키 : value, 키는 인덱스로
series = pd.Series({'사과':1200, '딸기':3500,'키위':2100})
# print(series)

series.name = "과일가격"
print(series)

# 인덱스 수정 
# 인덱스 수정은 전체 수정은 가능하나, 일부 부분 수정은 불가능
# series.index[0] = '오렌지' 이렇게 수정하는 건 안됨
series.index = ['오렌지','딸기','참외']
print(series)

# 인덱스 삭제
# 인덱스 삭제는 불가능함, 그러나 초기화는 가능
series = series.reset_index(drop=True)
print(series)
```
