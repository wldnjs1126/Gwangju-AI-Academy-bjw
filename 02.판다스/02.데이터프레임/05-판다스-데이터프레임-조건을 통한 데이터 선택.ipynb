```python
import pandas as pd
import numpy as np

data = {
    '이름' : ['유재석', '박명수', '정준하', '노홍철', '정형돈', '하하'],
    '지역' : ['서울', '부산', '부산', '서울', '서울', '서울'],
    '나이' : [19, 23, 20, 25, 18, 21],
    '국어점수' : [86, 90, 80, 65, 50, 60],
    '수학점수' : [86, 100, 66, 70, 40, 80],
    '코딩' : ['Python', 'Java', np.nan, 'Javascript', 'PYTHON', np.nan]
}

df = pd.DataFrame(data=data,index=['1번','2번','3번','4번','5번','6번'])
print(df)
```

```python
# 불린 인덱싱
# 나이가 20살보다 큰 사람 정보를 뽑아내시오
print(df[df["나이"]>= 20])

print(df['나이']>=20)
```

```python
# 나이가 20살 이하인 사람을 뽑으시오
print(df[df["나이"]<= 20])
print(df[~(df["나이"]> 20)]) # ~ 은 not과 같은의미 
```

```python
# 수학 점수가 70점 이상인 데이터를 뽑아내시오
print(df[df["수학점수"]>= 70])
print(df[df["수학점수"]>= 70]['수학점수'])


# loc 활용하기
print(df.loc[df["수학점수"]>= 70,"수학점수"]) # 뒤에 따로 다시 행이름을 적어주면 해당 값만 출력됨
print(df.loc[df["수학점수"]>= 70,["수학점수",'지역']])
```

```python
# 지역이 부산이고 나이가 20살 이상
print(df.loc[(df["나이"]>= 20) & (df["지역"] =="부산")])
```

```python
# 나이가 20살 이상 또는 코딩이 Python인 사람
print(df.loc[(df["나이"]>= 20) & (df["코딩"] =="Python")])
```

```python
# 기본적으로 str로 나와야 하는데 타입이 object로 나왔을 때
# df["코딩"].str. <= 코딩에 해당하는 값을 문자열로 바꿔줌 - 문자열로 바꾸면 사용할 수 있는 함수가 생김 (upper이나 lower 등)
print(df.loc[(df["코딩"].str.upper() =="PYTHON")])

#      이름   지역  나이  국어점수  수학점수   코딩
# 1번  유재석  서울  19     86      86     Python
# 5번  정형돈  서울  18     50      40     PYTHON
```
