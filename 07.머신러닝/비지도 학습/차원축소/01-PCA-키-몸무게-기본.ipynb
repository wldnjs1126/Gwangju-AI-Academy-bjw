```python
# PCA는 차원을 줄이는 기법
# 여려 feature를 "정보 손실 최소화"로 압축
# 시각화 및 전처리에 많이 사용 : 4차원 이상의 데이터는 눈으로 볼 수 없지만 PCA를 통해 2차원이나 3차원으로 줄이면 그래프로 시각화하여 패턴을 읽음
# 다중공선성 해결에도 많이 사용
# 과적합 방지 : 머신 러닝 모델에 너무 많은 변수를 넣으면 모델이 복잡해져 성능이 떨어짐 , 핵심 변수 (주성분)만 추출해 학습 효율을 높임

# 제 1 주성분의 한계
# 데이터를 가장 잘 설명할 수는 있지만 모든 데이터를 설명할 수는 없음

# 제 2 주성분
# 데이터가 2번째로 많이 퍼진 방향
# 제 1 주성분이 설명하지 못하는 부분을 커버

# 데이터가 N 차원 데이터라면 제 N 차원 주성분까지 생김, 그러나 2차원이기에 제 2 주성분까지만.

# 어떻게 구할까?
# PCA의 가장 큰 철학은 많이 퍼져 있는 방향 = 데이터를 잘 설명하는 특성 

# 공분산 행렬의 고유벡터와 고유 값
# 고유 벡터 : 제일 넓게 퍼진 방향 (음의 방향인지, 양의 방향인지)
# 고유 값 : 데이터가 얼마나 퍼져 있는지, 그 주성분이 얼마나 설명력이 있는지 
# 제일 큰 고유값을 갖는 고유 벡터가 제 1 주성분임

# 두 변수 X,Y 사이에 선형관게가 있는가를 알아보는 게 공분산
# 공분산이 음수이면 X가 커짐에 따라 Y는 반대로 작아지는 경향이 있음

######## 상관관계 그래프 이름 - 히트맵 관련해서 시험 문제 나옴 #######
```

```python
# PCA (Principal Component Analysis) => 주성분 분석

# Principal => 주요한, 핵심적인
# Component => 구성 요소 
# Analysis => 분석

import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

X = np.array([
    [170, 65],
    [175, 70],
    [180, 75],
    [185, 80]
])

pca = PCA(n_components=1)

# 1. fit() => 데이터의 규칙만 학습,데이터 자체는 변경 하지 않음
# 2. transform() => 학습한 규칙으로 실제 변환(새로운값 생성)
# X_scaled = scaler.transform(X)
# 3. fit_transform() → fit + transform 한 번에 실행

x_pca = pca.fit_transform(X)

print("원본데이터 : \n",X)
print("PCA 결과 : \n",x_pca)

plt.figure()
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```
