```python
# 선형 회귀의 이해
# 원인과 결과의 관게를 선으로 그리는 작업
# 원인이 되는 데이터 (X) VS 결과를 나타내는 데이터 (Y)
# 마법의 직선을 찾는 과정임 (가장 훌륭한 예측선 찾기)

# 선형 회귀는 주어진 데이터로부터, 관계를 가장 잘 나타내는 직선을 그리는 일
```

```python
# 1. 회귀란 무엇인가?

import pandas as pd

df = pd.DataFrame({
    "Study" : [1,2,3,4,5],
    "Score" : [55,65,70,80,90],
})

df
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study"],df["Score"])

plt.xlabel("Study Time")
plt.ylabel("Score")

plt.show()
```

```python
# 선형회귀 모델 만들기
from sklearn.linear_model import LinearRegression

# pip install scikit-learn

x = df[["Study"]]  # df 뒤에 [] 하나는 1차원 시리즈로, [[]] 두개는 2차원 데이터 프레임으로 구현됨
y = df["Score"]

print(x)
print(y)

# 학습
# 1. 평균 계산
# 2. 기울기 계산
# 3. 절편 계산
# 4. model.coef_ 저장
# 5. model.intercept_ 저장

model = LinearRegression()
model.fit(x,y)  # fit 함수는 뒤 x,y 값을 넣어주면 알아서 직선을 찾아줌, 모델한테 데이터를 주면서 (최적의 직선 경로) 공부,학습 시키는 함수

print(model.coef_)  # 기울기
print(model.intercept_)  # 절편

pred = model.predict([[6]])  # 데이터 입력시에는 2차원 배열 형태로 적어야 함
print(pred)
```

```python
df = pd.DataFrame({
    "size":[20,25,30,35,40],
    "price":[200,230,280,310,350]
})

x = df[["size"]] 
y = df["price"]

print(x)
print(y)

model = LinearRegression()
model.fit(x,y) 

print(model.coef_) 
print(model.intercept_)

pred = model.predict([[45]])
print(pred)
```

```python
# 총 결제 금액으로 팁 예측하기

import seaborn as sns
from sklearn.linear_model import LinearRegression

# 데이터 불러오기
tips = sns.load_dataset("tips")

# X(독립변수), y(종속변수)
X = tips[['total_bill']]
y = tips['tip']

# 모델 생성 및 학습
model = LinearRegression()
model.fit(X, y)

# total_bill = 40일 때 팁 예측
pred_tip = model.predict([[40]])

print("예상 팁:", pred_tip[0])
```

```python
# 인원이 많을수록 팁은 증가하는가?

import seaborn as sns
from sklearn.linear_model import LinearRegression

# 데이터 불러오기
tips = sns.load_dataset("tips")

# X(독립변수), y(종속변수)
X = tips[['size']]
y = tips['tip']

# 모델 생성 및 학습
model = LinearRegression()
model.fit(X, y)

# 인원 6명일 때 팁 예측
pred_tip = model.predict([[7]])

print("예상 팁:", pred_tip[0])
```
