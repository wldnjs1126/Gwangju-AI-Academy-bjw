```python
import pandas as pd

df = pd.DataFrame({
    "study":[1,2,3,4,5,6,7,8],
    "sleep":[8,7,7,6,6,5,5,4],
    "homework":[2,4,6,8,10,12,14,16],
    "score":[52,58,65,70,76,82,90,95]
})

print(df)
```

```python
# 2. 회귀 모델 만들기

import statsmodels.api as sm

x = df[["study","sleep","homework"]]
y = df["score"]

x = sm.add_constant(x)  # 1단계. 절편 추가
model = sm.OLS(y,x)  # 오차 제곱합을 최소로 하는 직선을 찾겠다는 뜻

result = model.fit()

#  R-squared(R²) : 0.999 거의 1 

print(result.summary())

# P>|t| : 변수 3개가 의미가 있는지 없는지 확인 
# 귀무가설 : 의미가 없다
# 대립가설 : 의미가 있다

# P>|t|
# study  p- value : 0.000  => p < 0.05 => 대립가설 : 의미가 있음
# sleep p - value : 0.178  => p > 0.05 => 귀무가설 : 의미가 없음
# homework p - value : 0.000  => p < 0.05 => 대립가설  : 의미가 있음    # 이렇게 해석 가능

# 다중공선성 문제
```

```python
# 다중공선성

# 입력 변수(X)들끼리 너무 비슷해서 
# 모델이 어떤 변수가 진짜 영향을 주는지 구분하기 어려운 상태 
# score= 5×study + 2×lecture
# score= 100×study − 93×lecture
import pandas as pd

df = pd.DataFrame({
    "study":[1,2,3,4,5,6,7,8],
    "sleep":[8,7,7,6,6,5,5,4],
    "score":[52,58,65,70,76,82,90,95]
})

print(df)

# | 공부시간 | 강의시청시간 | 점수 |

# |    1 |    1.1 | 50 |
# |    2 |    2.2 | 60 |
# |    3 |    3.1 | 70 |
# |    4 |    4.2 | 80 |
```

```python
import statsmodels.api as sm

x = df[["study","sleep"]]
y = df["score"]

x = sm.add_constant(x)
model = sm.OLS(y,x)
result = model.fit()
print(result.summary())
```
