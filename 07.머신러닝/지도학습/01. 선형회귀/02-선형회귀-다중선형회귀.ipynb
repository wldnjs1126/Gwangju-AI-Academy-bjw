```python
# 다중 선형회귀

import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "study": [1,2,3,4,5,6,7,8],
    "sleep": [8,7,7,6,6,5,5,4],
    "homework": [1,1,2,2,3,3,4,4],
    "score": [50,55,60,65,72,78,85,90]  # 상단 3개는 독립 변수, score는 종속 변수
})

# score = (w1 * study) + (w2 * sleep) + (w3 * homework) + b 
# w 는 가중치로 기울기를 의미하는데, 현재 w1 ~ w3까지 다중의 기울기를 가지고 있어서 다중 선형회귀라고 함

x = df[["study","sleep","homework"]]
y = df["score"] # 정답 = 실제 값

model = LinearRegression()
# x => (행, 열) : 2차원
# y => (행,) : 1차원
model.fit(x,y)

print(model.coef_)  # [ 3.91666667 -1.33333333  2.58333333]
print(model.intercept_)  # 53.29166666666668

# score = (3.91 * study) + (-1.33 * sleep) + (2.58 * homework) + 53.29
# 1시간 공부를 하면 3.91점이 올라가고, 1시간 잠을 자면 1.33점 내려가고, 숙제를 1개 하면 2.58점 올라감

# 공부 6시간
# 수면 6시간
# 과제 3개

pred = model.predict([[6,6,3]])  # [76.54166667]
print(pred)
```
