```python
# KNN : 최근접 이웃
# KNN 알고리즘의 개념
# KNN 알고리즘은 판별하고 싶은 데이터와 인접한 K 개수의 데이터를 찾아, 해당 데이터의 라벨이 다수인 범주로 데이터를 분류하는 방식
# 예를 들어 K = 3 라 하고 내 주변에 짜장 매니아 2과 짬뽕 매니아 1명이 있다면 나는 짜장 매니아

# 주의점으로는 변수 크기 차이가 너무 많이 나면 판단하기에 어려움이 있으니 표준화 시킨다음에 진행하는 것이 좋음

from sklearn.neighbors import KNeighborsClassifier

x = [[1], [2], [3], [4]]  # 공부 시간
y = [0,0,1,1]  # 합격 / 불합격

model = KNeighborsClassifier(n_neighbors=3)  # 3,5,7 홀수로..

model.fit(x,y)

result = model.predict([[2.5]])
print(result)
```
