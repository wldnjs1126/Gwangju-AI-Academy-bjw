```python
import seaborn as sns
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    classification_report
)
```

```python
# 1. 데이터 수집 

titanic = sns.load_dataset("titanic")
titanic
```

```python
# 2. 필요한 컬럼 선택

df = titanic[["survived","age", "fare", "sex", "pclass"]].copy()  # 원본에 영향을 주지 않게 복사
df
```

```python
# 3. 결측치 처리

# df.info()  # age : 714 non-null => 결측치가 많아서 채워야 함

df["age"] = df["age"].fillna(df["age"].median())
```

```python
# 4. one hot encoding

df = pd.get_dummies(df,columns=["sex"],drop_first=True)

df["sex_male"] = df["sex_male"].astype(int)
df

# one hot encoding (원핫인코딩)

# 주로 범주형 데이터를 숫자형 데이터로 바꾸는데 사용, 원핫인코딩은 범주형 데이터를 순서나 크기 정보 없이 0과 1로 표현하는 방법
# 함수 이름 get_dummies로 뒤에 원하는 컬럼명을 적으면 예를 들어 성별에서는 남자와 여자가 한 그룹으로 되어있는데 이를 남성, 여성으로 나눠서 표현함

# 예를 들면 모니터, 마우스, 키보드 중에서 고르라고 하면 범주형으로 데이터 작성이 되는데
# 이를 데이터 저장에 편하게 숫자를 하나씩 대입시켜서 각 1,2,3 로 칭할 수 있지만 그렇게 되면 계산식에서 1 + 2 = 모니터 + 마우스 인데 키보드 값이 나옴
# 즉, 연속적인 데이터가 아니라서 오류가 발생 따라서 만약 범주형이 3개 라면 3개의 칸을 두고 순서 대로 [1 0 0], [0 1 0] ,[0 0 1]이렇게 저장을 하면
# 오류 발생도 일어나지 않고 계산에도 편함 => 이게 원핫인코딩임
```

```python
# 5. x, y 분리 (독립 변수, 종속 변수)

x = df[["age","fare", "pclass", "sex_male"]]
y = df["survived"]
```

```python
# 6. 학습용 / 테스트용 데이터 분리

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=100
)
```

```python
# 7. 모델 생성

model = LogisticRegression(max_iter=1000)

# 8. 학습

model.fit(X_train,y_train)

# 9. 예측

y_pred = model.predict(X_test)

# 10. 평가
print("정확도")
print(accuracy_score(y_test,y_pred))

print("분류리포트")
print(classification_report(y_test,y_pred))
```

```python
# 11. 결과 지표

import statsmodels.api as sm

# -------------------------
# 1. X, y 준비
# -------------------------
X = df[["age", "fare", "pclass", "sex_male"]]
y = df["survived"]

# -------------------------
# 2. 상수항 추가 (필수)
# -------------------------
X_sm = sm.add_constant(X)

# -------------------------
# 3. Logit 모델
# -------------------------
model = sm.Logit(y, X_sm)
result = model.fit()

# -------------------------
# 4. 결과 출력
# -------------------------
print(result.summary())
```
