````python
import pandas as pd

df = pd.DataFrame({
    'study_time': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'grade':      [0, 0, 0, 1, 1, 1, 2, 2, 2]
})

df
````

````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

x= df[["study_time"]]
y = df["grade"]

model.fit(x,y)

pred = model.predict(x)
print(pred)  # [0 0 0 1 1 1 2 2 2]

# 공부 시간이 10시간일 때, 4.5시간일 때

pred = model.predict([[10],[4.5]])
print(pred)

print("클래스", model.classes_)
print("확률",model.predict([[10]]))
````

# 로지스틱 회귀(Logistic Regression) 정리

## 1. 로지스틱 회귀 개념

| 항목 | 내용 |
|------|------|
| 정의 | 데이터를 분류(Classification)하기 위한 머신러닝 알고리즘 |
| 목적 | 특정 클래스에 속할 확률을 예측 |
| 결과 | 0 또는 1 (이진 분류) |
| 활용 | 합격/불합격, 스팸/정상메일, 질병 유무 |
| 특징 | 선형회귀 결과를 시그모이드 함수로 변환 |

---

## 2. 선형회귀와 로지스틱 회귀 비교

| 구분 | 선형회귀 | 로지스틱 회귀 |
|------|----------|--------------|
| 목적 | 숫자 예측 | 분류 |
| 출력값 | 연속형 값 | 확률(0~1) |
| 예시 | 집값, 점수 예측 | 합격 여부, 스팸 여부 |
| 함수 | 직선 | 시그모이드 함수 |
| 손실함수 | MSE | Log Loss(Cross Entropy) |

---

## 3. 시그모이드 함수

| 항목 | 내용 |
|------|------|
| 역할 | 선형 결과를 0~1 사이 확률로 변환 |
| 출력 범위 | 0 ~ 1 |
| 특징 | S자 형태 그래프 |
| 의미 | 결과를 확률로 해석 가능 |

예시

공부시간 → 합격확률

1시간 → 0.1

5시간 → 0.5

10시간 → 0.9

---

## 4. 로지스틱 회귀 동작 과정

| 단계 | 설명 |
|------|------|
| 1 | 입력값(X) 수집 |
| 2 | 선형 계산 수행 |
| 3 | 시그모이드 함수 적용 |
| 4 | 확률 계산 |
| 5 | 기준값(0.5)으로 분류 |

---

## 5. 경사하강법(Gradient Descent)

| 항목 | 내용 |
|------|------|
| 목적 | 손실함수 최소화 |
| 방법 | 가중치를 반복적으로 수정 |
| 핵심 | 오차가 줄어드는 방향으로 이동 |
| 반복 | 예측 → 오차계산 → 가중치수정 |

업데이트 공식

새 가중치 = 기존 가중치 - (학습률 × 기울기)

---

| predict_proba() | 클래스별 확률 반환 |
