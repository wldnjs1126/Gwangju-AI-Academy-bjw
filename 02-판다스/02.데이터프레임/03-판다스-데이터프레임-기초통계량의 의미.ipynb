```python
import pandas as pd

# 데이터 생성
scores = pd.Series([80, 82, 83, 84, 85, 86, 87, 88, 89, 90])

# 기본 통계
print(scores.describe())

# 평균, 중앙값, 분산, 표준편차 순으로 중요하게 봄
# 1. 평균이 높고 표준 편차가 작음 => 전체적으로 우수, 고른 형태
# 2. 평균이 높고 표준편차가 큼 => 개인간의 차이가 매우 큼, 상위권과 하위권 분리 현상, 성과 양극화 상태
# 3. 평균이 낮고 표준편차 작음 => 전체 점수가 낮음, 학생 간의 차이가 작음, 난이도가 높은 상태
# 4. 평균과 중앙 값 차이가 큼 => 이상치 (일부 극단 값 - 유난히 높거나 낮은 값) 존재
# 5. 사분위수가 좁음 => 데이터 안전성 높음, 변동 폭이 낮음
# 6. 사분위수가 넓음 => 학생들 간의 격차가 큰 상태, 변동성이 높음
# 7. 표준편차 큼, 사분위수 좁음 => 이상치 의심, 데이터 입력 오류 가능성
# Q1 - 25%, Q2 - 50%, Q3 - 75% / IQR = Q3-Q1 , IQR 값이 작으면 그 사이 50% 범위에 엄청 빼곡하게 있다라는 의미

# describe() 실무 해석 순서
# 평균 - 중앙값 - 표준편차 - 사분위수 - 최대/최소값 - 이상치 여부 판단 - 최종 결론 도출

# 통계량은 숫자 자체보다 데이터의 특징과 상태를 설명하는 해석도구, 
# 실제 분석 목표는 숫자 계산이 아니라 숫자를 의미있는 스토리로 변환하는 과정
```
