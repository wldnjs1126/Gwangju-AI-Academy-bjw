```python
import pandas as pd
import numpy as np

# 딕셔너리로 생성 시 자동으로 키가 시리즈의 인덱스로 들어감
# 딕셔너리의 구조는 키 : value, 키는 인덱스로
series = pd.Series({'사과':1200, '딸기':3500,'키위':2100})
print(series)

# 단일값 접금
print(series['사과'])
print(series['딸기']) # 결과 값이 스칼라

print(series[ ['딸기','사과'] ]) # 팬시 인덱싱

# 과일 중에 2000원 이상의 과일을 시리즈로 뽑아내시오
print(series[series>=2000]) # 시리즈로 뽑아줌
```

```python
# 행번호 인덱싱
print(series[1:3])
print(series[0]) # 단일 값
print(series[:])

# 인덱스 기반 인덱싱 = 행번호 기반
print(series['사과' : '키위'])
print(series['사과' : '딸기']) # 설정된 범위만 보여줌

```

```python
# iloc 함수 활용 i = integer = 행번호 기반 
print(series.iloc[0:3]) # 함수
print(series.iloc[1:3]) # 함수, 인덱스 0인 사과를 제외하고 2개만 나옴

# 특정 위치에 있는 값 가져오기
print(series.iloc[0]) # 인덱스 0 인 사과의 값 1200

# 시리즈 추가 # loc를 사용해도 되고 안해도 됨
# 없으면 추가 , 있으면 업데이트
# 인덱스 이름으로
series.loc['바나나']=5000 
print(series)

series['참외']=6000
print(series)

# iloc는 행 번호를 기반으로 하고 loc는 인덱스 이름을 기반으로 함
# loc 같은 경우 해당 인덱스 이름이 없다면 추가를 해주고 있다면 업데이트를 함, 예를 들어 바나나가 없으면 추가 되는 것

# 업데이트

series['참외'] = 7000
print(series)

series.iloc[-1] = 9000 # 참외가 맨 마지막에 들어갔으므로 -1
print(series)

# 처음에는 참외 7000이 출력되었다가 업데이트가 되고 나면 9000으로 나옴
# iloc도 업데이트 가능
```

```python
# 인덱스 삭제
series.drop('사과')
print(series)
# 사과를 삭제하려고 했지만 삭제가 되지 않음

series.drop('사과',inplace=True)
print(series)
# 뒤에 inplace를 적어주면 삭제가 됨

series_2 = series.drop('사과')
print(series_2)
# 사과가 삭제된 시리즈를 시리즈2라는 새로운 이름을 붙인 형태
```
