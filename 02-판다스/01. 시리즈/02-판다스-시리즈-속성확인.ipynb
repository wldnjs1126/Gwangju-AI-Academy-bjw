```python
import pandas as pd
import numpy as np

# 딕셔너리로 생성 시 자동으로 키가 시리즈의 인덱스로 들어감
# 딕셔너리의 구조는 키 : value, 키는 인덱스로
series = pd.Series({'사과':1200, '딸기':3500,'키위':2100})
print(series)

# 속성확인
print(series.index) # Index(['사과', '딸기', '키위'], dtype='object') => 리스트로 관리됨
print(series.values) # [1200 3500 2100] => 넘파이 배열로 나옴

# 넘파이 공통
print(series.dtype) # int64
print(series.shape) # (3,)
print(series.size) # 3
```
