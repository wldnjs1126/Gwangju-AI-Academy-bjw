```python
# 매틀 플롯 그래프 그리는 방법은 크게 2가지로 설명하고 있음
# 객체 기반 스타일로 그래프 그리기

# 1. 기본으로 그리기 = 기본으로 제공되는 기반으로 그리기
# 2. 객체 기반으로 그리기 = 커스텀 마이징
```

```python
import matplotlib.pyplot as plt

x = [ i for i in range(1,6)]
y = [ i for i in range(1,6)]

plt.plot(x,y)
plt.title("첫번째 그래프")
plt.xlabel("x축")
plt.ylabel("y축")

#plt.show()
```

```python
# 객체 기반으로 그리기
# fig는 전체 도화지를 의미함
# ax는 그래프를 그릴 수 있는 곳
# fig는 스케치북 그 자체라고 보면 ax는 그 안에 있는 도화지

x = [1,2,3,4,5]
y = [1,2,3,4,5]

# plt 안에서 fig와 ax 객체가 뽑아져 나와서 그 객체를 토대로 그래프를 그림
fig, ax = plt.subplots()
ax.plot(x,y)
ax.set_xlabel("x축")
ax.set_ylabel("y축")

fig.set_dpi(300) # dpi는 1인치당 픽셀
fig.savefig("first_plot.png")
```

```python
# Figure와  Axes 2개 생성 (1행 2열 구조)
fig , ax = plt.subplots(1,2,figsize=(10,4)) # 가로 10인치 세로 4인치
ax[0].plot([1,2,3,4],[1,2,3,4],color="blue",marker="o")
ax[0].set_title("first")
ax[0].set_xlabel("X axis")
ax[0].set_ylabel("Y axis")

ax[1].plot([1,2,3,4],[1,2,3,4],color="yellow",marker="x")
ax[1].set_title("first")
ax[1].set_xlabel('X-axis',loc="right",labelpad=15, fontdict={'family': 'serif', 'color': 'b', 'weight': 'bold', 'size': 14})
ax[1].set_ylabel("Y axis")

# 그래프는 각각의 코드가 존재하므로 모두 같은 양식이 아니라 해당 라벨이나 이름에 새로운 값을 주면 해당 객체만 수정이 가능
```

```python
# 라벨 관련 옵션

plt.plot([1,2,3,4],[1,4,6,9])
plt.xlabel('X-axis',loc="right",labelpad=15, fontdict={'family': 'serif', 'color': 'b', 'weight': 'bold', 'size': 14})
plt.ylabel('Y-axis',loc="top",labelpad=20, fontdict={'family': 'serif', 'color': 'b', 'weight': 'normal', 'size': "xx-large"})
# labelpad는 라벨과 그래프와의 간격을 의미
```

```python
# 범례 legend

plt.plot([1,2,3,4],[1,4,6,9],label="price")

# 범례 위치 지정
# plt.legend(loc=(0.5,0.5))
plt.legend(loc='lower right')
```

```python
# 눈금선 표시 컨트롤
import numpy as np

x = np.arange(1, 11)

y1 = x ** 2
y2 = x ** 3
y3 = x * 4

plt.plot(x, y1,'ro--', x, y2, 'gs--', x, y3, 'b^--')
plt.legend(['red', 'green', 'blue'], loc='best')

# 눈금선의 스타일 설정(옵션 설정)
# describes a sequence of 5 point and 1 point dashes separated by 2
# "길이가 5인 선, 그 다음 간격 2, 길이가 1인 선, 그 다음 간격 2"의 패턴이 반복되도록 설정
# which='both' = 주눈금선(major ticks)과 보조눈금선(minor ticks) 모두를 표시

plt.grid(which="major",axis='x',color="green",alpha=0.5,dashes=(5,2,1,2))
plt.grid(which="major",axis='y',color="red",alpha=0.5,linestyle='-.',linewidth=1)
```

```python
names = ['a','b','c']
values = [50, 30, 100]

plt.subplot(121) # 1행 2열의 subplot 중 왼쪽 (1번) 위치
plt.bar(names,values)

plt.subplot(122) # 1행 2열의 subplot 중 왼쪽 (2번) 위치
plt.barh(names,values)
```
